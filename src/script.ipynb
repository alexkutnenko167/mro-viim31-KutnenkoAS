```python
import numpy
import matplotlib.pyplot as plt

# Функция для считывания данных из файла
def read_data():
    with open('data.txt', 'r') as file:
        lines = file.readlines()
        
    data = []
    for i in range(0, len(lines), 3):  # Проходим по строкам с шагом 3 (2 строки + 1 пропуск)
        if i + 1 < len(lines):  # Проверяем, что есть еще строки
            x_coords = list(map(int, lines[i].split()))
            y_coords = list(map(int, lines[i + 1].split()))
            data.append((x_coords, y_coords))
    
    return data
```

```python
import numpy as np
import matplotlib.pyplot as plt

    # Функция для чтения данных из файла
def read_data_ho(file_path, file_path2):
    X, y, weights, bias = [],[],[], 0
    
    with open(file_path, 'r') as file:
        for line in file:
            values = list(map(int, line.strip().split()))
            X.append(values[:-1])       # Все кроме последнего значения
            y.append(values[-1])        # Последнее значение - метка

    with open(file_path2, 'r') as file:
        values = file.read().replace('\n','')
        values = list(map(float, values.strip().split()))
        weights.append(values[0])
        weights.append(values[1])
        bias = values[2]
    return np.array(X), np.array(y), np.array(weights), bias

def k_means(ak, ak2):

    with open('data3.txt', 'r') as file:
        values = file.read().replace('\n','')
        values = list(map(float, values.strip().split()))
        ak.append(values[0])
        ak.append(values[1])
        ak2.append(values[2])
        ak2.append(values[3])
    print("Центр класса 1",ak)
    print("Центр класса 2",ak2)
    return ak, ak2
```

```python
def get_ready(center1, center2):
    data = read_data()

    # Цвета для графиков
    colors = ['#FF4100', '#FFCA00', '#7908AA', '#39142F', '#3357FF', '#3357FF']

    # Построение графиков
    for i, (x, y) in enumerate(data):
        plt.scatter(x, y, color=colors[i % len(colors)], label=f'Класс {i + 1}')

        min_x = min(x)
        max_x = max(x)
        min_y = min(y)
        max_y = max(y)

        plt.fill([min_x, max_x, max_x, min_x, min_x], [min_y, min_y, max_y, max_y, min_y],
                color=colors[i % len(colors)], linewidth = 0, alpha = 0.2)


    plt.scatter(center1[0], center1[1], marker = 'x', color = "blue", s = 25, label = "Центр класса 1 (k-means)")
    plt.scatter(center2[0], center2[1], marker = 'x', color = "black", s = 25, label = "Центр класса 2 (k-means)")

    plt.title('График с элементами классов')
    plt.xlabel('X координаты')
    plt.ylabel('Y координаты')
    plt.legend()
    plt.grid()
    plt.show()
```

```python
# Чтение данных
X, y, weights, bias = read_data_ho('data2.txt', 'weights_bias.txt')

for i in np.unique(y):
  plt.scatter(X[y == i][:, 0], X[y == i][:, 1], color=colors[i % len(colors)], label=f'Класс {i + 1}')

print()

# Линия разделения
x_values = np.linspace(min(X[:, 0]/2), max(X[:, 0]/2), 2)
y_values = -(weights[0] * x_values + bias) / weights[1]

plt.plot(x_values, y_values, color='red')
plt.xlabel('X координаты')
plt.ylabel('Y координаты')
plt.title('Алгоритм Хо-Кашьяпа')
plt.legend(["Класс 1", "Класс 2", "Гиперплоскость (Хо-Кашьяп)"])
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def read_data(filename):
    X = []
    y = []
    with open(filename, 'r') as f:
        for line in f:
            line = line.strip().split()
            if line:
                x = [float(val) for val in line[:-1]]
                X.append(x)
                y.append(int(line[-1]))
    return np.array(X), np.array(y)

colors = ['#FF4100', '#FFCA00', '#7908AA', '#39142F', '#3357FF', '#3357FF'] # цвета

X, y = read_data("data_perceptron.txt")
for i in np.unique(y):
  plt.scatter(X[y == i][:, 0], X[y == i][:, 1], color=colors[i % len(colors)], label=f'Класс {i + 1}',
               edgecolors='k') # покраска точек

if X.shape[1] != 2:
    print("Визуализация поддерживается только для двумерных данных.")
else:
    plt.xlabel("Признак 1")
    plt.ylabel("Признак 2")
    plt.legend(["Класс 1", "Класс 2"])
    plt.title("Данные")
    plt.grid()
    plt.show()
```
